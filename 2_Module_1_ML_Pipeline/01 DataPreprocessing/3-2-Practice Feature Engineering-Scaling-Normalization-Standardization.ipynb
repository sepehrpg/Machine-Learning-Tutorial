# Simple Implement Each one of Scaling Types

```python
# 0: Import Dataset
import pandas as pd
import numpy as np
df = pd.read_csv('../../../../../0-Dataset_And_Playground/1-AI-Dataset/housing1.csv')
df.head(n=10)
```

```python
# 01 Just Select Number
df = df.select_dtypes(include=np.number)
df.head(n=10)
```

```python
# 1. Absolute Maximum Scaling
# - np.max(np.abs(df), axis=0): Calculates the maximum absolute value for each column.
# - df / max_abs: Divides each value by the maximum absolute value of its column to scale the data.
# - scaled_df.head(): Displays the first few rows of the scaled dataset.

max_abs = np.max(np.abs(df), axis=0)
scaled_df = df / max_abs
scaled_df.head(n=10)
```

```python
# 2. Min-Max Scaling
# - MinMaxScaler(): Creates a scaler object for Min-Max scaling.
# - scaler.fit_transform(df): Calculates min and max values and scales the dataset between 0 and 1.

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head(n=10)
```

```python
# 3. Normalization (Vector Normalization)
# - Normalizer(): Creates a normalizer object to scale data.
# - scaler.fit_transform(df): Normalizes each row so its vector length becomes 1.

from sklearn.preprocessing import Normalizer
scaler = Normalizer()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head(n=10)
```

```python
# 4. Standardization
# - standardScaler(): Creates a scaler for standardizing the data.
# - scaler.fit_transform(df): Subtracts the mean and divides by the standard deviation.

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data,columns=df.columns)
scaled_df.head(n=10)
```

```python
# 5. Robust Scaling
# - RobustScaler(): Creates a scaler that uses median and IQR for scaling.
# - scaler.fit_transform(df): Scales the data while reducing the influence of outliers.

from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data,columns=df.columns)
scaled_df.head(n=10)
```

```python
# 6. L1 Normalization (Manhattan / Taxicab Normalization)
# - Normalizer(norm='l1'): Creates a normalizer object that uses the L1 norm.
# - scaler.fit_transform(df): Normalizes each row so that the sum of absolute values = 1.
#   This means for each sample (row):
#     X_scaled = X_i / sum(|X_j|)  for all j in that row.
# - scaled_df.head(): Shows the first few rows of the L1-normalized dataset.

from sklearn.preprocessing import Normalizer
scaler = Normalizer(norm='l1')
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head(n=10)
```

```python
# 7. L2 Normalization (Euclidean Normalization)
# - Normalizer(norm='l2'): Creates a normalizer object that uses the L2 norm (default in sklearn).
# - scaler.fit_transform(df): Normalizes each row so that the Euclidean length (L2 norm) = 1.
#   This means for each sample (row):
#     X_scaled = X_i / sqrt(sum(X_j^2))  for all j in that row.
# - scaled_df.head(): Shows the first few rows of the L2-normalized dataset.
from sklearn.preprocessing import Normalizer
scaler = Normalizer(norm='l2')   # norm='l2' is also the default
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head(n=10)
```
